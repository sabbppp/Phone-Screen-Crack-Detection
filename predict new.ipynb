```python
import os
import numpy as np
import matplotlib.pyplot as plt

from tensorflow.keras.models import load_model
from tensorflow.keras.utils import load_img, img_to_array
```

```python
BASE_DIR = r"C:\Users\ASUS\Desktop\Phone_screen CNN"

MODEL_PATH = os.path.join(BASE_DIR, "phone_screen_model.keras")
PREDICT_DIR = os.path.join(BASE_DIR, "predict")
```

```python
model = load_model(MODEL_PATH)

print("Model loaded successfully!")
```

```python
images = os.listdir(PREDICT_DIR)

image_path = os.path.join(PREDICT_DIR, images[0])

print("Image selected:", images[0])
```

```python
img = load_img(image_path, target_size=(128, 128))

img_array = img_to_array(img)
img_array = img_array / 255.0
img_array = np.expand_dims(img_array, axis=0)

prediction = model.predict(img_array)

if prediction[0][0] >= 0.5:
    result = "Cracked Phone Screen"
else:
    result = "Normal Phone Screen"

print("Prediction:", result)
```

```python
plt.imshow(load_img(image_path))
plt.title(result)
plt.axis("off")
plt.show()
```

```python
plt.imshow(load_img(image_path))
plt.title(result)
plt.axis("off")
plt.show()
```

```python
for image in images:

    image_path = os.path.join(PREDICT_DIR, image)

    img = load_img(image_path, target_size=(128, 128))
    img_array = img_to_array(img) / 255.0
    img_array = np.expand_dims(img_array, axis=0)

    prediction = model.predict(img_array, verbose=0)

    if prediction[0][0] >= 0.5:
        result = "Cracked"
    else:
        result = "Normal"

    print(image, ":", result)

    plt.imshow(load_img(image_path))
    plt.title(result)
    plt.axis("off")
    plt.show()
```
